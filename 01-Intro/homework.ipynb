#### Q7. Sum of weights

```python
X = df[df['origin'] == 'Asia'][['vehicle_weight', 'model_year']][:7].values
```

```python
X
```

```python
X.shape
```

```python
X_T = X.T
```

```python
X_T
```

```python
XTX = X_T.dot(X)
```

```python
np.linalg.inv(XTX)
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
np.linalg.inv(XTX).dot(X_T).dot(y).sum()
```
